```python
import os
import pandas as pd
```

```python
labels_brset = pd.read_csv(os.path.join(os.getcwd(), 'labels_brset.csv'))
canon_df = labels_brset[labels_brset['camera'] == 'Canon CR']
nikon_df = labels_brset[labels_brset['camera'] == 'NIKON NF5050']
canon_ids = canon_df['patient_id'].unique()
nikon_ids = nikon_df['patient_id'].unique()
overlap_ids = set(canon_ids) & set(nikon_ids)
print(overlap_ids)
overlap_camera_df = labels_brset[labels_brset['patient_id'].isin(overlap_ids)]
# ten patients had images taken by both cameras
```

```python
no_diabetes_ids = labels_brset[labels_brset['diabetes'] == 'No'].unique()
diabetes_ids = labels_brset[labels_brset['diabetes'] == 'Yes'].unique()
overlap_diabetes_ids = set(no_diabetes_ids) & set(diabetes_ids)
```

```python
canon_diabetes_df = canon_df[canon_df['diabetes'] == 'yes']
canon_diabetes_ids = canon_diabetes_df['patient_id'].unique()
nikon_diabetes_df = nikon_df[nikon_df['diabetes'] == 'yes']
nikon_diabetes_ids = nikon_diabetes_df['patient_id'].unique()
overlap_diabetes_ids = set(nikon_diabetes_ids) & set(overlap_ids)
# among the people with diabetes, no patients had images taken by both cameras
```

```python
canon_no_diabetes_df = canon_df[canon_df['diabetes'] == 'No']
canon_no_diabetes_ids = canon_no_diabetes_df['patient_id'].unique()
overlap_dm_ids = set(canon_no_diabetes_ids) & set(canon_diabetes_ids)
canon_overlap_dm_df = canon_df[canon_df['patient_id'].isin(overlap_dm_ids)]
# among images taken by the Canon camera, 3 patients had images taken with and without diabetes
```

```python
canon_0_ids = canon_df[canon_df['DR_ICDR'] == 0]['patient_id'].unique()
canon_nonpro_ids = canon_df[canon_df['DR_ICDR'].isin([1, 2, 3])]['patient_id'].unique()
canon_pro_lds = canon_df[canon_df['DR_ICDR'] == 4]['patient_id'].unique()
overlap_0_nonpro_ids = set(canon_0_ids) & set(canon_nonpro_ids)
overlap_nonpro_pro_ids = set(canon_nonpro_ids) & set(canon_pro_lds)
overlap_0_pro_ids = set(canon_0_ids) & set(canon_pro_lds)
# among images taken by the Canon camera, 63 patients had images graded with normal and non-proliferative DR
# among images taken by the Canon camera, 17 patients had images graded with normal and proliferative DR
# among images taken by the Canon camera, 29 patients had images graded with non-proliferative and proliferative DR
```
